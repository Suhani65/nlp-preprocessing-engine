## Task 1: Conceptual Understanding
This section explains key NLP concepts like case sensitivity, stopwords, and stemming vs lemmatization.

1. "Love" vs "love":
They are treated as different words due to case sensitivity. Lowercasing makes them the same.

2. Without removing stopwords:
Model gets noisy and less accurate.

3. When removing stopwords is harmful:
- Sentiment analysis ("not good" → "good")
- Search queries ("to be or not to be")

4. Stemming vs Lemmatization:
Stemming cuts words (running → runn)  
Lemmatization gives proper root (running → run)


```python
import re
from collections import Counter
```

## Task 2: Text Preprocessing
Here, we build a function to clean and normalize raw text by removing noise and generating meaningful tokens.

```python
def preprocess_text(text):
    if not text or not isinstance(text, str):
        return [], ""

    text = text.lower()
    text = re.sub(r"http\S+|www\S+|@\S+", "", text)
    text = re.sub(r"\d+", "", text)
    text = re.sub(r"(.)\1{2,}", r"\1", text)
    text = re.sub(r"[^a-z\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()

    tokens = text.split()
    tokens = [w for w in tokens if len(w) > 2 or w in ["no", "not"]]

    clean_sentence = " ".join(tokens)

    return tokens, clean_sentence
```

## Task 3: Stress Testing
We test the preprocessing function on diverse noisy sentences to evaluate its robustness.

```python
sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]
```

```python
all_tokens = []

for s in sentences:
    tokens, clean = preprocess_text(s)

    print("Original:", s)
    print("Tokens:", tokens)
    print("Clean Sentence:", clean)
    print("-" * 50)

    all_tokens.extend(tokens)
```

## Task 4: Token Analytics
This section analyzes token statistics like count, uniqueness, and average length.

```python
for s in sentences:
    tokens, clean = preprocess_text(s)

    avg_len = sum(len(w) for w in tokens) / len(tokens) if tokens else 0

    print("Sentence:", s)
    print("Total tokens:", len(tokens))
    print("Unique tokens:", len(set(tokens)))
    print("Avg token length:", avg_len)
    print("-" * 50)
```

## Task 5: Frequency Analysis
We identify the most and least frequent words using token frequency counts.

```python
counter = Counter(all_tokens)

print("Top 10 words:", counter.most_common(10))
print("Least 5 words:", counter.most_common()[-5:])
```

## Task 6: Full Pipeline
We combine all preprocessing steps into a single pipeline for handling multiple inputs efficiently.

```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens, clean = preprocess_text(text)
        all_tokens.extend(tokens)
        clean_sentences.append(clean)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }
```

## Task 7: Error Handling
We test edge cases like empty input and invalid data to ensure the function is robust.

```python
test_cases = ["", "😂😂😂", "123456"]

for t in test_cases:
    print("Input:", t)
    print("Output:", preprocess_text(t))
    print("-" * 30)
```
